But how does this work? How can computer code (at its base just a bunch of `0` and `1`), which is inherently deterministic, generate completely random numbers?

### Pseudo-Random Number Generators

A Pseudo-Random number generator is a deterministic map $f$ from $[0,1)$ to $[0,1)$, i.e.
\begin{equation}
x_{n+1} = f(x_{n})
\end{equation}
that has zero autocorrelation for which $P(x)$ is the uniform distribution (i.e., all values between in the interval $[0,1)$ are equally probable).

All you need to know is computer scientists have developed very efficient versions of these algorithms, which, provided some initial "**seed**" value $x_{0}$, can generate long sequences of pseudo-random numbers $\{x_{n}\}$ for $n \in [1, N]$.

The reason I call these *pseudo*-random is because the entire sequence can be deterministically computed if the seed $x_{0}$ is exactly known.

### "True" randomness

If all computer programs used the same seed and algorithm, the "random" sequence of numbers they generate would in fact be all exactly identical and thus not random at all.

The surprising workaround is that, when a computer program is launched, it computes the random seed based on the *truly random* fluctuations of the real world (as experienced by the computer hardware), such as the temperature of the Compute Processing Unit (CPU) used by the program.

### Reproducible randomness

As computational scientists, it is occasionally very useful for us to have control over these pseudo-random sequences of numbers. This can be done by over-writing the truly random seed with a deterministically-specified seed.

```julia
Random.seed!(1)
Random.rand(5)
```

```julia
Random.seed!(1)
Random.rand(5)
```

Different seeds give different answers:

```julia
Random.seed!(99)
Random.rand(5)
```

```julia
Random.seed!(99)
Random.rand(5)
```
